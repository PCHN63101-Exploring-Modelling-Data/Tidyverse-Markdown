# Introduction: Tidyverse & RMarkdown

This week builds on your base R foundations by introducing tools for a clearer, more efficient data-analysis workflow. You will learn to use packages from the tidyverse to reshape data, create and transform variables, calculate grouped summaries, and produce effective visualisations with `ggplot2`.

Base R remains an important part of your R toolkit: it provides the core language, data structures, and problem-solving skills that underpin everything you do in R. The tidyverse is not a replacement for those skills, but a collection of packages that makes common analysis tasks easier to express and read, particularly when those tasks involve complex series of steps. We will end this week by introducing R Markdown, so you can combine code, results, graphics, and written interpretation into reproducible HTML and PDF reports.


As you work through this week's content, you can try out the code snippets in your sandbox. Head over to posit cloud here: https://posit.cloud/spaces/797313/content/
