
Now let's suppose I only want the rows that relate to Adelie and Gentoo penguins:

```r
my_penguins <- filter(my_penguins, species == 'Adelie' | species == 'Gentoo')
```
Again, very straightforward and the logical operators work in exactly the same way as we have seen in earlier weeks. Given there are only three species in the data, a more efficient way of achieving the same thing would be to specify what we don't want:

```r
my_penguins <- filter(my_penguins,! species == 'Chinstrap')
```


## Creating and editing columns

To add or edit a column, we can use the `mutate()` function from `dplyr`. Suppose we wanted to add a column that recorded the ratio of our penguins's bill lengths to their flipper lengths.

```r
penguins <- mutate(penguins,
                   billFlipperRatio = bill_len/flipper_len)

head(penguins)
```
```text
  species    island bill_len bill_dep flipper_len body_mass    sex year billFlipperRatio
1  Adelie Torgersen     39.1     18.7         181      3750   male 2007        0.2160221
2  Adelie Torgersen     39.5     17.4         186      3800 female 2007        0.2123656
3  Adelie Torgersen     40.3     18.0         195      3250 female 2007        0.2066667
4  Adelie Torgersen       NA       NA          NA        NA   <NA> 2007               NA
5  Adelie Torgersen     36.7     19.3         193      3450 female 2007        0.1901554
6  Adelie Torgersen     39.3     20.6         190      3650   male 2007        0.2068421
```
Because `billFlipperRatio` didn't exist in the data frame, `mutate()` created it. We could just as easily create a new column with logical values:

```r
penguins <- mutate(penguins,
                   longFlipper = flipper_len > 213)
```
```text
  species    island bill_len bill_dep flipper_len body_mass    sex year billFlipperRatio longFlipper
1  Adelie Torgersen     39.1     18.7         181      3750   male 2007        0.2160221       FALSE
2  Adelie Torgersen     39.5     17.4         186      3800 female 2007        0.2123656       FALSE
3  Adelie Torgersen     40.3     18.0         195      3250 female 2007        0.2066667       FALSE
4  Adelie Torgersen       NA       NA          NA        NA   <NA> 2007               NA          NA
5  Adelie Torgersen     36.7     19.3         193      3450 female 2007        0.1901554       FALSE
6  Adelie Torgersen     39.3     20.6         190      3650   male 2007        0.2068421       FALSE
```

 Now imagine we want the `body_mass` column to be in kilos rather than grams. 


```r
penguins <- mutate(penguins,
                   body_mass = body_mass/1000)

head(penguins)
```
```text
  species    island bill_len bill_dep flipper_len body_mass    sex year billFlipperRatio longFlipper
1  Adelie Torgersen     39.1     18.7         181      3.75   male 2007        0.2160221       FALSE
2  Adelie Torgersen     39.5     17.4         186      3.80 female 2007        0.2123656       FALSE
3  Adelie Torgersen     40.3     18.0         195      3.25 female 2007        0.2066667       FALSE
4  Adelie Torgersen       NA       NA          NA        NA   <NA> 2007               NA          NA
5  Adelie Torgersen     36.7     19.3         193      3.45 female 2007        0.1901554       FALSE
6  Adelie Torgersen     39.3     20.6         190      3.65   male 2007        0.2068421       FALSE
```

In this case, the `body_mass` column did already exist so it was overwritten with each value now divided by 1000.

## Grouping and Summarising

Grouping and summarising often go together because we want to see summaries according to some grouping variable or other. To tell R what groupings exist in a data frame we use the `group_by()` function. It has no visible effect on a data frame but it affects how subsequent functions interact with it. Let's look at an example. If we wanted the mean penguin `body_mass` we could use the `summarise()` function:

```r
summarise(penguins,
          meanWeight = mean(body_mass, na.rm = TRUE))
```
```text
# A tibble: 1 × 1
  meanWeight
       <dbl>
1       4.20
```
This gives us the mean body mass across all penguins. Now let's see what happens if we first specify a grouping:

```r
penguins <- group_by(penguins, species)

summarise(penguins,
          meanWeight = mean(body_mass, na.rm = TRUE))

```

```text
# A tibble: 3 × 2
  species   meanWeight
  <fct>          <dbl>
1 Adelie          3.70
2 Chinstrap       3.73
3 Gentoo          5.08
```
`summarise()` now behaves differently because we have specified a grouping variable, in this case `species`. Once you have specified a group using `group_by()` any subsequent functions applied to that data frame will be applied *per group*.

What makes this a really useful pair of functions is that you can specify as many summary statistics and groups as you like:

```r
penguinSummary <- group_by(penguins,species,island)

penguinSummary <- summarise(penguinSummary,
                            meanWeight = mean(body_mass, na.rm = TRUE),
                            sdWeight = sd(body_mass, na.rm = TRUE),
                            maxWeight = max(body_mass, na.rm = TRUE),
                            nPenguins = n()
)

print(penguinSummary)
```

```text
# A tibble: 5 × 6
# Groups:   species [3]
  species   island    meanWeight sdWeight maxWeight nPenguins
  <fct>     <fct>          <dbl>    <dbl>     <dbl>     <int>
1 Adelie    Biscoe          3.71    0.488      4.78        44
2 Adelie    Dream           3.69    0.455      4.65        56
3 Adelie    Torgersen       3.71    0.445      4.7         52
4 Chinstrap Dream           3.73    0.384      4.8         68
5 Gentoo    Biscoe          5.08    0.504      6.3        124
```
Now we have a nice tibble showing each of our summary stats for each level of our two grouping variables.

````{tip}
`group_by()` and `summarise()` would often be piped together, so the above code would become:

```r
penguinSummary <- penguins %>% 
  group_by(species,island) %>%
  summarise(meanWeight = mean(body_mass, na.rm = TRUE),
            sdWeight = sd(body_mass, na.rm = TRUE),
            maxWeight = max(body_mass, na.rm = TRUE),
            nPenguins = n())
```
````
