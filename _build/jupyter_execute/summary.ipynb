# Summary

This week, you learned how to extend your base R skills using the tidyverse. You used `tidyr` to reshape data with `pivot_longer()`, `dplyr` to create and modify variables with `mutate()`, and calculate group-level summaries with `group_by()` and `summarise()`. You also created clear, informative visualisations with `ggplot2`.

You learned that tidyverse tools work with the same underlying R objects and concepts introduced in earlier weeks. Base R knowledge remains useful for understanding data types, handling missing values, inspecting objects, and troubleshooting code.

Finally, you learned how to write reproducible reports with R Markdown. You learned about explanatory narrative in Markdown, including R code and its output, and knitting documents to PDF.
