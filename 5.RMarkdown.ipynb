A key responsibility of a researcher or data scientist is communicating their findings and the analyses that support them. You may need to explain an analytical process to collaborators, prepare a report for your employer, or write a manuscript for submission to a scientific journal.

You could simply share your R script, but code alone is often difficult to interpret, even for someone who is proficient in R. A better approach is to create a document that combines explanatory text with your R code and its outputs. This provides context for the analysis, explains how results should be interpreted, and makes the work easier to follow.

R Markdown makes this possible by combining executable R code, the output it generates, and clearly formatted text in a single document. In R Markdown terminology, this explanatory text is called *narrative*.

The web page you are reading now is written in Markdown.

## Narrative and comments

Narrative text tells the story of an analysis. It explains the question being addressed, why a particular approach was chosen, the data and methods used, the main findings, and what those findings mean. It should give readers enough context to understand and evaluate the work without requiring them to read every line of code.

Code comments, written with the `#` symbol, serve a different purpose. They help someone read, debug, or update the code. For example, a comment might explain why certain records were excluded, why missing values were handled in a particular way or what a complex calculation is doing.

Use the report narrative to explain the analytical rationale and communicate results to the intended audience. Use comments to clarify parts of the code that would otherwise be confusing. Avoid repeating the same explanation in both places; instead, let each serve its distinct role.


## Formatting text in R Markdown

In RStudio you can start an R Markdown document by clicking on the new file button in the top left of the interface and selecting R Markdown. A pop-up box will then ask you to enter a title and select the format of your final report. We will mostly be using pdf, but you can also produce web pages or Word documents. Once you click OK, you will be presented with a template R Markdown document. Lines 1 to 6 contain the information you entered in the initial pop-up box and can be edited if needed. The rest of the document is there to provide some examples of how RMarkdown works, you can safely delete it. 

When the Markdown document is processed RStudio will look for certain special characters that tell it how to format the text. A single # creates a level 1 header, ## creates a level 2 header and so on. 

````markdown
# Header 1
## Header 2
````
Gets rendered as

> # Header 1
> ## Header 2

Placing ** either side of a word makes that word bold. A single * either side makes the word italicised.

````Markdown
**bold**

*italicised*
````
gets rendered as:

> **bold**

> *italicised*

When you want to include a variable name or the name of a function it is customary to put a backtick either side of it so that it gets formatted as code.

````markdown
`mean()`
````
gets rendered as

> `mean()`

Note that the apostrophe ' and the backtick ` are not the same thing. Only backticks will format your text as code.


## Inserting R code into R Markdown

To include R code in your markdown document you need to add an R 'chunk'. You do this by typing:
````markdown

```{r}

```
````
Anything you write in between the opening and closing ``` gets interpreted as R code. Again, those are backticks not apostrophes.

Inside the brackets that start an R chunk you can specify parameters that affect what happens when that chunk is run.


| Option | Effect |
|---|---|
| `echo = FALSE` | Hide the code; show output. |
| `eval = FALSE` | Display the code but do not run it. |
| `include = FALSE` | Run the code but show neither code nor output. |
| `results = "hide"` | Run code but hide printed output. |
| `warning = FALSE` | Suppress warnings in the rendered document. |
| `message = FALSE` | Suppress package/startup messages in the rendered document. |
| `error = TRUE` | Show errors in the rendered document and continue rendering. |
| `fig.width = 7, fig.height = 5` | Set plot dimensions. |

So if you wanted a chunk of code that suppressed messages and warnings from the console, you would write:

````markdown

```{r message = FALSE, warning = FALSE}

```
````
If you don't set any parameters, the defaults mean that your code will be run, and the output will be displayed along with any warnings and messages.

## Knitting the document

Once you have written your markdown and chunks of R code, it is time to produce your pdf (or html or Word doc). You do this by pressing the knit button on the toolbar at the top of the pane.

After a few seconds a preview will pop up of your pdf file (if your browser settings allow for it). The pdf will also appear in the files tab of output pane so you have access to it whenever you need it.

If the resulting document is not to your liking, you can of course edit your markdown file and knit it again.
